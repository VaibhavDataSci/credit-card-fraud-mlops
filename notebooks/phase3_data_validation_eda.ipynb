# Phase 3 — Data Validation & Exploratory Data Analysis (EDA)

This notebook performs automated data validation checks and exploratory data analysis on the raw Credit Card Fraud dataset (`data/raw/AIML DATASET.csv`).

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yaml

# Ensure project root is in path
sys.path.insert(0, '..')
from src.data.validation import DataValidator

sns.set_theme(style='whitegrid')
```

## 1. Automated Data Validation

```python
with open('../params.yaml', 'r') as f:
    params = yaml.safe_load(f)

val_config = params.get('data_validation', {})
val_config['raw_data_path'] = '../data/raw/AIML DATASET.csv'

validator = DataValidator(val_config)
report = validator.validate()
print('Overall Validation Status:', report['overall_status'])
report['checks']
```

## 2. Load Raw Dataset for EDA

```python
df = pd.read_csv('../data/raw/AIML DATASET.csv')
print(f'Dataset Shape: {df.shape[0]:,} rows, {df.shape[1]} columns')
df.head()
```

## 3. Class Imbalance Analysis

```python
counts = df['isFraud'].value_counts()
print('Non-Fraud (0):', counts[0])
print('Fraud (1):', counts[1])
print(f'Fraud Ratio: {counts[1]/len(df)*100:.3f}%')
```
